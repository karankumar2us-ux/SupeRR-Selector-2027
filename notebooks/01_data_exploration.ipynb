```python

# BATSMAN PERFORMANCE ANALYSIS

# Group deliveries by player ID
batting_stats = df.groupby(
    ["striker_id", "striker_name"]
).agg(
    total_runs=("runs", "sum"),
    fours=("is_four", "sum"),
    sixes=("is_six", "sum"),
    dismissals=("is_wicket", "sum"),
    total_deliveries=("runs", "count")
).reset_index()

# Sort players by runs
batting_stats = batting_stats.sort_values(
    "total_runs",
    ascending=False
)

print("TOP 20 BATSMEN:")
display(batting_stats.head(20))
```

```python

# CORRECTED BATTING PERFORMANCE

# Create accurate delivery-level indicators
df["is_ball_faced"] = (
    df["extras_type"].str.lower() != "wide"
)

df["batter_dismissed"] = (
    (df["is_wicket"] == True) &
    (df["dismissed_player_id"] == df["striker_id"])
)

# Aggregate player statistics
batting_stats = df.groupby(
    ["striker_id", "striker_name"]
).agg(
    total_runs=("runs", "sum"),
    balls_faced=("is_ball_faced", "sum"),
    fours=("is_four", "sum"),
    sixes=("is_six", "sum"),
    dismissals=("batter_dismissed", "sum")
).reset_index()

# Calculate batting strike rate
batting_stats["strike_rate"] = (
    batting_stats["total_runs"] /
    batting_stats["balls_faced"]
) * 100

# Calculate boundary runs percentage
batting_stats["boundary_runs"] = (
    batting_stats["fours"] * 4 +
    batting_stats["sixes"] * 6
)

batting_stats["boundary_percentage"] = (
    batting_stats["boundary_runs"] /
    batting_stats["total_runs"].replace(0, float("nan"))
) * 100

# Sort by total runs
batting_stats = batting_stats.sort_values(
    "total_runs",
    ascending=False
)

print("CORRECTED TOP 20 BATSMEN:")
display(
    batting_stats.head(20).round(2)
)
```

```python

# PLAYER-WISE MATCH PERFORMANCE

match_batting = df.groupby(
    ["match_id", "striker_id", "striker_name"]
).agg(
    runs=("runs", "sum"),
    balls_faced=("is_ball_faced", "sum"),
    dismissals=("batter_dismissed", "sum"),
    fours=("is_four", "sum"),
    sixes=("is_six", "sum")
).reset_index()

# Calculate match-wise strike rate
match_batting["strike_rate"] = (
    match_batting["runs"] /
    match_batting["balls_faced"].replace(0, float("nan"))
) * 100

# Aggregate match-wise performances by player
consistency_stats = match_batting.groupby(
    ["striker_id", "striker_name"]
).agg(
    matches_played=("match_id", "nunique"),
    average_runs=("runs", "mean"),
    median_runs=("runs", "median"),
    highest_score=("runs", "max"),
    lowest_score=("runs", "min"),
    consistency_std=("runs", "std"),
    scores_30_plus=("runs", lambda x: (x >= 30).sum()),
    scores_50_plus=("runs", lambda x: (x >= 50).sum())
).reset_index()

# Sort by average runs
consistency_stats = consistency_stats.sort_values(
    "average_runs",
    ascending=False
)

print("TOP 20 CONSISTENT BATTERS:")
display(consistency_stats.head(20).round(2))
```

```python

# FILTER PLAYERS WITH AT LEAST 10 APPEARANCES

reliable_batters = consistency_stats[
    consistency_stats["matches_played"] >= 10
].copy()

# Rank players by average runs
reliable_batters = reliable_batters.sort_values(
    "average_runs",
    ascending=False
)

print("TOP 20 BATTERS (MINIMUM 10 APPEARANCES):")

display(
    reliable_batters.head(20).round(2)
)
```

```python

# CALCULATE STANDARD CRICKET BATTING AVERAGE

player_batting_average = match_batting.groupby(
    ["striker_id", "striker_name"]
).agg(
    total_runs=("runs", "sum"),
    total_dismissals=("dismissals", "sum"),
    innings=("match_id", "nunique")
).reset_index()

player_batting_average["batting_average"] = (
    player_batting_average["total_runs"] /
    player_batting_average["total_dismissals"].replace(0, float("nan"))
)

# Require at least 10 batting appearances
player_batting_average = player_batting_average[
    player_batting_average["innings"] >= 10
]

player_batting_average = player_batting_average.sort_values(
    "batting_average",
    ascending=False
)

print("TOP BATTERS BY CRICKET BATTING AVERAGE:")

display(
    player_batting_average.head(20).round(2)
)
```
